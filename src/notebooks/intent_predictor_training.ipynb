```python
import pandas as pd, joblib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
```

```python
df = pd.read_csv("../../data/commands.csv")
df.head()
```

```python
print(df.info())
print("*"*40)
uniques = df.intent.unique()
print(f"total intents: {len(uniques)}")
print(uniques)
```

```python
vectorizer = TfidfVectorizer()
vect_commands = vectorizer.fit_transform(df["command"])

label_encoder = LabelEncoder()
encoded_intents = label_encoder.fit_transform(df["intent"])
```

```python
X, y = vect_commands, encoded_intents

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=22
)

model = LinearSVC()
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, predictions))
print(classification_report(y_test, predictions))
```

```python
def predict_intent(text):
    X = vectorizer.transform([text])
    scores = model.decision_function(X)[0]    
    pred_idx = scores.argmax()
    pred_label = label_encoder.inverse_transform([pred_idx])[0]
    confidence = scores[pred_idx]              # not real probability, but ranking

    return pred_label, float(confidence)
```

```python
predict_intent("restart immediatly")
```

```python
joblib.dump(model, "../../models/intent_predictor_model.pkl")
joblib.dump(vectorizer, "../../models/vectorizer.pkl")
joblib.dump(label_encoder, "../../models/label_encoder.pkl")
```
